# Day 1 – House Price Predictor (Linear Regression)

This notebook trains a Linear Regression model to predict house prices based on the size of the house (in sqft). It uses simple historical data for demonstration.

```python
import numpy as np #numbers
import matplotlib.pyplot as plt #plotting
from sklearn.linear_model import LinearRegression #linear regression
```

```python
#CREATE DATASET
 # X = square footage, y = price in $1000s
X = np.array([[500], [800], [1000], [1200], [1500]])
y = np.array([150, 200, 250, 275, 300])

'''
We are teaching the model:
500 sqft = $150K
800 sqft = $200K
1000 sqft = $250K
1200 sqft = $275K
1500 sqft = $300K
The model will learn this pattern
'''
```

```python
# TIME TO TRAIN THE MODEL
model = LinearRegression() #create the model
model.fit(X, y) #fit the model to the data

'''
This is where training happens. The model finds the line that best fits the data points using least squares.
The model is now trained and can be used to make predictions.
The model has learned the relationship between square footage and price.'''
```

```python
#Time to make predictions
new_sizes = np.array([[1100],[1600]])
predicted_prices = model.predict(new_sizes)

print("Predicted prices for 1100 and 1600 sqft:", predicted_prices)
```

```python
#VIsualize the results
plt.scatter(X, y, color='blue', label='Training data') #plot the training data

#plot the regression line
plt.plot(X, model.predict(X), color='red', label='Regression line') #plot the regression line
plt.xlabel("House Size (sqft)") #x-axis label
plt.ylabel("Price ($1000s)") #y-axis label
plt.title("House Price Prediction") #title of the plot
plt.legend() #show the legend
plt.grid() #show the grid
plt.show() #show the plot
'''
The blue dots are the training (actual) data points
The red line is the regression line that the model has learned (prediction logic).
The model can now predict the price of a house based on its size.
The model has learned the relationship between square footage and price.
The model can now predict the price of a house based on its size.'''
```

```python
#Peek inside the model
print("Model coefficient (slope):", model.coef_[0])
print("Model intercept:", model.intercept_)

'''
The model coefficient (slope) tells us how much the price increases for each additional square foot.
The model intercept tells us the price when the size is 0 sqft (not very meaningful in this case).'''
```
